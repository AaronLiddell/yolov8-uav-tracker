```python
#Updating Data
df.loc[df["CustomerName"] == "Anna Ivanova"] #returns data frame with Customer Name = Anna
```

```python
df.loc[df["CustomerName"] == "Anna Ivanova", "Product"] = "Tim" #changes product to Tim
df.loc[df["CustomerName"] == "Anna Ivanova"]
```

```python
df.loc[df["Country"] == "USA", "Country"] = "United States" #change all USA to United States
df.loc[df["Country"] == "United States"]
```

```python
df["Country"] = df["Country"].str.upper() #all countries in caps
df["Country"]
```

```python
#Deleting Data
df = df.drop(39) #removes row 39
```

```python
#Cleaning Data
df.dropna() #drop anything containing null values
df.fillna({"OrderID": 0}, inplace=True] #fills any empty OrderID with 0, and modifies existing dataframe, rather than creating new
```

```python
df.rename(columns={"OrderID": "Order ID"}, inplace=True) #renames OrderID column
```

```python
#Analyzing Data
df["Country"].value_counts() #number of times each country appears in df
```

```python
df.groupby("Country")["Price"].sum() #total price per country
```
